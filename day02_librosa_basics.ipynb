```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Librosa 내장 샘플 오디오 불러오기
y, sr = librosa.load(librosa.ex('trumpet'))
```

```python
print(f"오디오 길이: {len(y)} 샘플")
print(f"샘플레이트: {sr} Hz")
print(f"재생 시간: {len(y)/sr:.2f}초")
```

```python
plt.figure(figsize=(12, 4))
librosa.display.waveshow(y, sr=sr)
plt.title("파형 (Waveform)")
plt.xlabel("시간 (초)")
plt.ylabel("진폭")
plt.show()
```
