```python
from gtts import gTTS
from IPython.display import Audio
import os

# 텍스트를 음성으로 변환
text = "I like to eat apples and oranges"
tts = gTTS(text=text, lang='en', slow=False)
tts.save("output_tts.mp3")
print("생성 완료!")

# 재생
Audio("output_tts.mp3")
```

```python
sentences = [
    "The cat sat on the mat",
    "She sells seashells by the seashore",
    "How much wood would a woodchuck chuck"
]

for i, text in enumerate(sentences):
    tts = gTTS(text=text, lang='en', slow=False)
    tts.save(f"output_{i+1}.mp3")
    print(f"생성: {text}")
    display(Audio(f"output_{i+1}.mp3"))
```

```python
text = "The thirty three thieves thought that they thrilled the throne"

# 빠른 속도
tts_fast = gTTS(text=text, lang='en', slow=False)
tts_fast.save("fast.mp3")

# 느린 속도
tts_slow = gTTS(text=text, lang='en', slow=True)
tts_slow.save("slow.mp3")

print("빠른 속도:")
display(Audio("fast.mp3"))
print("느린 속도:")
display(Audio("slow.mp3"))
```
